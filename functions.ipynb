### Transformations => TRANSFORM

This is the tricky part — it’s not explicitly specified, but typically includes: stroke shortening, stroke merging, rotation (slight), distortion to fit layout, simplification.

>Think: wt(A, B) ≈ a(transform(A), B).

| Original | Meaning                | Normalize |
| -------- | ---------------------- | --------- |
| `w`      | weak composition       | TRANSFORM |
| `wt`     | transformed            | TRANSFORM |
| `wb`     | bottom transform       | TRANSFORM |
| `wr`     | rotation               | TRANSFORM |
| `wtr`    | transform-right        | TRANSFORM |
| `wtl`    | transform-left         | TRANSFORM |
| `wbl`    | bottom-left transform  | TRANSFORM |
| `wbr`    | bottom-right transform | TRANSFORM |
